```python
import numpy as np

# Load the employee dataset as a NumPy structured array.
employees = np.genfromtxt(
    "employee_salary_dataset.csv",
    delimiter=",",
    names=True,
    dtype=None,
    encoding="utf-8",
)

employees[:5]
```

# NumPy Arrays for Employee Salary EDA

This practical uses only NumPy arrays and array operations to study employee demographics, experience, roles, education, cities, and monthly salary.

```python
# Structured-array fields and numeric columns used throughout the analysis.
print("records:", employees.shape[0])
print("fields:", employees.dtype.names)

experience = employees["Experience_Years"]
age = employees["Age"]
salary = employees["Monthly_Salary"]
print("experience sample:", experience[:5])
print("salary sample:", salary[:5])
```

## 1. Arrays vs Python Lists
NumPy stores homogeneous values compactly and applies compiled operations to whole arrays. Python lists store references to Python objects, so NumPy arrays generally use less memory and run faster for numeric workloads.
## 2. Array Creation
Create arrays with the core constructors requested in the practical.


```python
import timeit

list_values = list(range(1_000_000))
array_values = np.arange(1_000_000)
list_time = timeit.timeit("sum(value * 2 for value in list_values)", globals=globals(), number=3)
array_time = timeit.timeit("np.sum(array_values * 2)", globals=globals(), number=3)
print(f"Python list memory estimate: {sum(value.__sizeof__() for value in list_values) / 1024**2:.2f} MB")
print(f"NumPy array memory: {array_values.nbytes / 1024**2:.2f} MB")
print(f"list time: {list_time:.4f}s | NumPy time: {array_time:.4f}s")
```

```python
zeros = np.zeros(5, dtype=int)
ones = np.ones(5, dtype=int)
sequence = np.arange(0, 20, 2)
percentile_grid = np.linspace(0, 100, 6)

print("zeros:", zeros)
print("ones:", ones)
print("arange:", sequence)
print("linspace:", percentile_grid)
```

```python
salary_array = np.asarray(salary)
print("shape:", salary_array.shape)
print("ndim:", salary_array.ndim)
print("dtype:", salary_array.dtype)
print("size:", salary_array.size)
print("itemsize:", salary_array.itemsize, "bytes")

print("first salary:", salary_array[0])
print("first five salaries:", salary_array[:5])
print("last five salaries:", salary_array[-5:])
```

```python
salary_matrix = salary_array[:20].reshape(4, 5)
salary_cube = salary_array[:24].reshape(2, 3, 4)

print("2D array:\n", salary_matrix)
print("2D row 1:", salary_matrix[1])
print("2D rows 1-2, columns 2-4:\n", salary_matrix[1:3, 2:5])
print("nD element [1, 2, 3]:", salary_cube[1, 2, 3])
print("nD first block:\n", salary_cube[0])
```

```python
# Element-wise operations and vectorization avoid explicit row-by-row loops.
annual_salary = salary_array * 12
salary_after_raise = salary_array * 1.08
experience_bonus = experience * 2_000
adjusted_salary = salary_array + experience_bonus

print("annual salary sample:", annual_salary[:5])
print("8% raise sample:", salary_after_raise[:5])
print("experience bonus sample:", experience_bonus[:5])
print("adjusted salary sample:", adjusted_salary[:5])
```

```python
# Broadcasting: add one department-level allowance to every employee in that department.
department_names = np.unique(employees["Department"])
department_allowances = np.arange(department_names.size) * 1_000
allowance_by_employee = department_allowances[
    np.searchsorted(department_names, employees["Department"])
]

salary_with_allowance = salary_array + allowance_by_employee
print("departments:", department_names)
print("allowances:", department_allowances)
print("broadcasted allowances:", allowance_by_employee[:10])
print("salary with allowance:", salary_with_allowance[:10])
```

```python
# Aggregation functions summarize the whole employee population.
print("salary sum:", np.sum(salary_array))
print("salary mean:", np.mean(salary_array))
print("salary minimum:", np.min(salary_array))
print("salary maximum:", np.max(salary_array))
print("salary standard deviation:", np.std(salary_array))
print("salary variance:", np.var(salary_array))

# Row-wise and column-wise summaries on a 2D salary matrix.
print("row sums:", np.sum(salary_matrix, axis=1))
print("column means:", np.mean(salary_matrix, axis=0))
print("row minimums:", np.min(salary_matrix, axis=1))
print("column maximums:", np.max(salary_matrix, axis=0))
```

```python
# Boolean indexing selects employees matching a condition.
high_salary = salary_array >= 100_000
experienced = experience >= 10
print("high-salary employees:", employees[high_salary][["Name", "Monthly_Salary"]])
print("experienced employees:", employees[experienced][["Name", "Experience_Years"]])
print("experienced high earners:", np.sum(experienced & high_salary))

# Fancy indexing selects chosen positions without a loop.
selected_positions = np.array([0, 4, 9, 14])
print("selected names:", employees["Name"][selected_positions])
print("selected salaries:", salary_array[selected_positions])
```

```python
# Random numbers with a fixed seed make simulations reproducible.
rng = np.random.default_rng(42)
performance_scores = rng.normal(loc=75, scale=8, size=employees.size)
performance_scores = np.clip(performance_scores, 0, 100)

simulation = rng.integers(0, 2, size=(employees.size, 3))
print("performance scores:", performance_scores[:10].round(2))
print("three simulation trials for first five employees:\n", simulation[:5])

# Re-seeding reproduces the same random sequence.
rng_again = np.random.default_rng(42)
print("same first score after reseeding:", rng_again.normal(75, 8))
```

## NumPy EDA Takeaways

- Use `shape`, `ndim`, `dtype`, `size`, and `itemsize` to understand an array.
- Use vectorized arithmetic, broadcasting, boolean indexing, and fancy indexing for concise employee analysis.
- Use axis `0` for column-wise summaries and axis `1` for row-wise summaries in a 2D array.
- Seed random generators when simulation results need to be reproduced.
