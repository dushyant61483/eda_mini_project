# Employee Salary Dataset: Exploratory Data Analysis

This notebook explores an employee salary dataset using **NumPy**, **Pandas**, and **Matplotlib**. We will inspect the data, perform numerical calculations, summarize groups, and create charts to understand salary patterns.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

file_name = "employee_salary_dataset.csv"
df = pd.read_csv(file_name)

print(df.head())
print("Rows and columns:", df.shape)
```

## 1. Load the Dataset

We begin by importing the required libraries and reading the CSV file into a Pandas DataFrame. The first rows and the shape of the DataFrame confirm that the file was loaded correctly and show how many records and columns are available.

```python
# Basic pandas inspection
print(df.columns)
print(df.dtypes)
print(df.isnull().sum())
print(df.describe())
```

## 2. Inspect the Data

These checks describe the dataset before analysis. We review the column names, data types, missing values, and summary statistics such as count, mean, minimum, and maximum.

```python
# NumPy arrays from employee columns
salary = np.array(df["Monthly_Salary"])
experience = np.array(df["Experience_Years"])
age = np.array(df["Age"])

print("Salary shape:", salary.shape)
print("Salary dimensions:", salary.ndim)
print("Salary data type:", salary.dtype)
print("Salary size:", salary.size)
print("Bytes per value:", salary.itemsize)

zeros = np.zeros(5)
ones = np.ones(5)
sequence = np.arange(0, 10, 2)
line_values = np.linspace(0, 100, 5)
print(zeros, ones, sequence, line_values)
```

## 3. Convert Columns to NumPy Arrays

Selected salary, experience, and age columns are converted to NumPy arrays. The array properties show the data structure, while the small examples demonstrate how NumPy creates zeros, ones, regular sequences, and evenly spaced values.

```python
# Indexing, slicing, reshaping, and transpose
print("First salary:", salary[0])
print("First five salaries:", salary[:5])

salary_table = salary[:20].reshape(4, 5)
print("Salary table:\n", salary_table)
print("First row:", salary_table[0])
print("Rows 2 and 3:\n", salary_table[1:3])
print("Transposed table:\n", salary_table.transpose())
print("Flattened table:", salary_table.flatten())
```

## 4. Indexing, Slicing, and Reshaping

This section accesses individual salary values and subsets of the array. The first 20 salaries are reshaped into a 4-by-5 table so we can compare rows, columns, the transposed layout, and the flattened one-dimensional form.

```python
# Element-wise calculations and broadcasting
annual_salary = salary * 12
salary_after_raise = salary * 1.10
salary_with_allowance = salary + 5000
salary_with_experience_bonus = salary + experience * 1000

print("Annual salary:", annual_salary[:5])
print("Salary after 10% raise:", salary_after_raise[:5])
print("Salary with allowance:", salary_with_allowance[:5])
print("Salary with experience bonus:", salary_with_experience_bonus[:5])
```

## 5. Element-wise Calculations

NumPy applies arithmetic to every employee's salary at once. We calculate annual salary, a 10% raise, a fixed allowance, and an experience-based bonus using vectorized operations and broadcasting.

```python
# NumPy summaries, axis operations, and filtering
print("Total salary:", np.sum(salary))
print("Average salary:", np.mean(salary))
print("Minimum salary:", np.min(salary))
print("Maximum salary:", np.max(salary))
print("Salary standard deviation:", np.std(salary))

print("Row totals:", np.sum(salary_table, axis=1))
print("Column averages:", np.mean(salary_table, axis=0))

high_salary = salary >= 100000
print("High salary employees:")
print(df[high_salary][["Name", "Department", "Monthly_Salary"]])

selected_rows = [0, 4, 9]
print("Selected employees:")
print(df.iloc[selected_rows][["Name", "City", "Monthly_Salary"]])
```

## 6. Summaries and Filtering

We calculate overall salary statistics and demonstrate NumPy axis operations on the reshaped table. Boolean filtering identifies employees earning at least 100,000 per month, while integer-location indexing selects specific rows.

```python
# Basic pandas grouping and sorting
average_salary_department = df.groupby("Department")["Monthly_Salary"].mean()
average_salary_education = df.groupby("Education_Level")["Monthly_Salary"].mean()
city_count = df["City"].value_counts()
education_count = df["Education_Level"].value_counts()

print("Average salary by department:")
print(average_salary_department)
print("\nAverage salary by education:")
print(average_salary_education)
print("\nEmployees by city:")
print(city_count)
print("\nEmployees by education:")
print(education_count)

print("\nTop five salaries:")
print(df.sort_values("Monthly_Salary", ascending=False).head())
```

## 7. Grouping, Counting, and Sorting with Pandas

Grouping helps compare average monthly salary across departments and education levels. We also count employees by city and education level, then sort the DataFrame to find the five highest-paid employees.

```python
# Reproducible random values for a simple employee score example
np.random.seed(42)
performance_score = np.random.randint(50, 101, size=len(df))
df["Performance_Score"] = performance_score
print(df[["Name", "Performance_Score"]].head())

np.random.seed(42)
second_scores = np.random.randint(50, 101, size=len(df))
print("Same random results:", np.array_equal(performance_score, second_scores))
```

## 8. Reproducible Random Performance Scores

A random performance score is added as an example of creating a new column. Setting the seed to 42 before each generation makes the results repeatable, which is important when testing or sharing analysis.

```python
# Chart 1: monthly salary distribution
plt.figure(figsize=(8, 4))
plt.hist(df["Monthly_Salary"], bins=8, color="skyblue", edgecolor="black")
plt.title("Monthly Salary Distribution")
plt.xlabel("Monthly salary")
plt.ylabel("Number of employees")
plt.show()
```

## 9. Visualize the Salary Distribution

The histogram shows how monthly salaries are distributed across the employees. The bars help reveal the most common salary ranges and whether the values are concentrated or spread out.

```python
# Chart 2: average salary by department
plt.figure(figsize=(8, 4))
average_salary_department.sort_values().plot(kind="bar", color="orange")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average monthly salary")
plt.xticks(rotation=0)
plt.show()
```

## 10. Compare Departments

This bar chart compares the average monthly salary for each department. Sorting the values before plotting makes the differences easier to read from lower to higher averages.

```python
# Chart 3: experience and salary
plt.figure(figsize=(8, 4))
plt.scatter(df["Experience_Years"], df["Monthly_Salary"], color="green")
plt.title("Experience and Monthly Salary")
plt.xlabel("Experience in years")
plt.ylabel("Monthly salary")
plt.grid(True)
plt.show()
```

## 11. Explore Experience and Salary

The scatter plot places years of experience on the horizontal axis and monthly salary on the vertical axis. A visible upward pattern would suggest that salary tends to increase with experience, although this chart alone does not prove causation.

```python
# Chart 4: employees by city
plt.figure(figsize=(8, 4))
city_count.sort_values().plot(kind="barh", color="purple")
plt.title("Number of Employees by City")
plt.xlabel("Number of employees")
plt.ylabel("City")
plt.show()
```

## 12. Count Employees by City

This horizontal bar chart shows how the employees are distributed across cities. It helps identify which locations have the largest and smallest representation in the dataset.

```python
# Chart 5: employees by education level
plt.figure(figsize=(8, 4))
education_count.plot(kind="bar", color="teal")
plt.title("Number of Employees by Education Level")
plt.xlabel("Education level")
plt.ylabel("Number of employees")
plt.xticks(rotation=0)
plt.show()
```

## 13. Count Employees by Education Level

This chart displays the number of employees in each education category. The result describes the composition of this sample and should not be treated as a description of the wider workforce.

```python
# Chart 6: average salary by education level
plt.figure(figsize=(8, 4))
average_salary_education.sort_values().plot(kind="bar", color="crimson")
plt.title("Average Salary by Education Level")
plt.xlabel("Education level")
plt.ylabel("Average monthly salary")
plt.xticks(rotation=0)
plt.show()
```

## 14. Compare Salary by Education Level

Here we compare the average monthly salary for each education level. This can reveal differences in the sample, but other factors such as department, experience, age, and location may also influence salary.

```python
# Chart 7: average salary by gender
average_salary_gender = df.groupby("Gender")["Monthly_Salary"].mean()

plt.figure(figsize=(6, 4))
average_salary_gender.plot(kind="bar", color=["steelblue", "salmon"])
plt.title("Average Salary by Gender")
plt.xlabel("Gender")
plt.ylabel("Average monthly salary")
plt.xticks(rotation=0)
plt.show()
```

## 15. Compare Salary by Gender

The final chart compares average monthly salary between gender categories in the dataset. It is a descriptive comparison only and should be interpreted carefully, especially because the sample size and other employee characteristics may differ between groups.

## Conclusion

This analysis demonstrated how to use NumPy and Pandas to inspect, transform, summarize, filter, group, and sort employee salary data. The visualizations provided views of salary distribution, average salary by department and education, the relationship between experience and salary, employee counts by city and education, and average salary by gender.

The results describe patterns in this dataset, but they do not establish cause-and-effect relationships. A stronger analysis would use a larger and more representative sample, check for outliers, calculate correlations or statistical tests, and consider additional factors before drawing broader conclusions.
