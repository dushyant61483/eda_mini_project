```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

file_name = "employee_salary_dataset.csv"
df = pd.read_csv(file_name)

print(df.head())
print("Rows and columns:", df.shape)
```

```python
# Basic pandas inspection
print(df.columns)
print(df.dtypes)
print(df.isnull().sum())
print(df.describe())
```

```python
# NumPy arrays from employee columns
salary = np.array(df["Monthly_Salary"])
experience = np.array(df["Experience_Years"])
age = np.array(df["Age"])

print("Salary shape:", salary.shape)
print("Salary dimensions:", salary.ndim)
print("Salary data type:", salary.dtype)
print("Salary size:", salary.size)
print("Bytes per value:", salary.itemsize)

zeros = np.zeros(5)
ones = np.ones(5)
sequence = np.arange(0, 10, 2)
line_values = np.linspace(0, 100, 5)
print(zeros, ones, sequence, line_values)
```

```python
# Indexing, slicing, reshaping, and transpose
print("First salary:", salary[0])
print("First five salaries:", salary[:5])

salary_table = salary[:20].reshape(4, 5)
print("Salary table:\n", salary_table)
print("First row:", salary_table[0])
print("Rows 2 and 3:\n", salary_table[1:3])
print("Transposed table:\n", salary_table.transpose())
print("Flattened table:", salary_table.flatten())
```

```python
# Element-wise calculations and broadcasting
annual_salary = salary * 12
salary_after_raise = salary * 1.10
salary_with_allowance = salary + 5000
salary_with_experience_bonus = salary + experience * 1000

print("Annual salary:", annual_salary[:5])
print("Salary after 10% raise:", salary_after_raise[:5])
print("Salary with allowance:", salary_with_allowance[:5])
print("Salary with experience bonus:", salary_with_experience_bonus[:5])
```

```python
# NumPy summaries, axis operations, and filtering
print("Total salary:", np.sum(salary))
print("Average salary:", np.mean(salary))
print("Minimum salary:", np.min(salary))
print("Maximum salary:", np.max(salary))
print("Salary standard deviation:", np.std(salary))

print("Row totals:", np.sum(salary_table, axis=1))
print("Column averages:", np.mean(salary_table, axis=0))

high_salary = salary >= 100000
print("High salary employees:")
print(df[high_salary][["Name", "Department", "Monthly_Salary"]])

selected_rows = [0, 4, 9]
print("Selected employees:")
print(df.iloc[selected_rows][["Name", "City", "Monthly_Salary"]])
```

```python
# Basic pandas grouping and sorting
average_salary_department = df.groupby("Department")["Monthly_Salary"].mean()
average_salary_education = df.groupby("Education_Level")["Monthly_Salary"].mean()
city_count = df["City"].value_counts()
education_count = df["Education_Level"].value_counts()

print("Average salary by department:")
print(average_salary_department)
print("\nAverage salary by education:")
print(average_salary_education)
print("\nEmployees by city:")
print(city_count)
print("\nEmployees by education:")
print(education_count)

print("\nTop five salaries:")
print(df.sort_values("Monthly_Salary", ascending=False).head())
```

```python
# Reproducible random values for a simple employee score example
np.random.seed(42)
performance_score = np.random.randint(50, 101, size=len(df))
df["Performance_Score"] = performance_score
print(df[["Name", "Performance_Score"]].head())

np.random.seed(42)
second_scores = np.random.randint(50, 101, size=len(df))
print("Same random results:", np.array_equal(performance_score, second_scores))
```

```python
# Chart 1: monthly salary distribution
plt.figure(figsize=(8, 4))
plt.hist(df["Monthly_Salary"], bins=8, color="skyblue", edgecolor="black")
plt.title("Monthly Salary Distribution")
plt.xlabel("Monthly salary")
plt.ylabel("Number of employees")
plt.show()
```

```python
# Chart 2: average salary by department
plt.figure(figsize=(8, 4))
average_salary_department.sort_values().plot(kind="bar", color="orange")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average monthly salary")
plt.xticks(rotation=0)
plt.show()
```

```python
# Chart 3: experience and salary
plt.figure(figsize=(8, 4))
plt.scatter(df["Experience_Years"], df["Monthly_Salary"], color="green")
plt.title("Experience and Monthly Salary")
plt.xlabel("Experience in years")
plt.ylabel("Monthly salary")
plt.grid(True)
plt.show()
```

```python
# Chart 4: employees by city
plt.figure(figsize=(8, 4))
city_count.sort_values().plot(kind="barh", color="purple")
plt.title("Number of Employees by City")
plt.xlabel("Number of employees")
plt.ylabel("City")
plt.show()
```

```python
# Chart 5: employees by education level
plt.figure(figsize=(8, 4))
education_count.plot(kind="bar", color="teal")
plt.title("Number of Employees by Education Level")
plt.xlabel("Education level")
plt.ylabel("Number of employees")
plt.xticks(rotation=0)
plt.show()
```

```python
# Chart 6: average salary by education level
plt.figure(figsize=(8, 4))
average_salary_education.sort_values().plot(kind="bar", color="crimson")
plt.title("Average Salary by Education Level")
plt.xlabel("Education level")
plt.ylabel("Average monthly salary")
plt.xticks(rotation=0)
plt.show()
```

```python
# Chart 7: average salary by gender
average_salary_gender = df.groupby("Gender")["Monthly_Salary"].mean()

plt.figure(figsize=(6, 4))
average_salary_gender.plot(kind="bar", color=["steelblue", "salmon"])
plt.title("Average Salary by Gender")
plt.xlabel("Gender")
plt.ylabel("Average monthly salary")
plt.xticks(rotation=0)
plt.show()
```
